# Exploratory EEG analysis

## Scientific question

What are the class balance, subject coverage and broad waveform differences for the selected decoding task?

## Rationale

This report checks the descriptive structure of the data without selecting subjects, tuning thresholds or fitting a classifier.

```python
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Markdown, display

from br_analysis.config import ProjectPaths
from br_analysis.data import load_epoch_dataset
from br_analysis.scientific_reports import cohort_table, plot_channel_effects, plot_class_waveforms, subject_trial_table
from br_analysis.tasks import get_task
```

```python
TASK_ID = "c_vs_ip"
paths = ProjectPaths.from_env("eda", create_dirs=False)
task = get_task(TASK_ID)
dataset = load_epoch_dataset(paths, task, "broadband")
```

```python
display(Markdown("## Selected task, classes and cohort"))
display(cohort_table(dataset))
display(subject_trial_table(dataset))
print(f"Subjects: {len(dataset.subject_ids)} | Trials: {len(dataset.y)} | Channels: {len(dataset.channel_names)}")
```

## Class and subject diagnostics

```python
counts = np.bincount(dataset.y, minlength=task.n_classes)
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(dataset.class_labels, counts)
ax.set(ylabel="Trials", title=f"Class availability: {task.id}")
plt.show()
subject_totals = subject_trial_table(dataset).set_index("subject").sum(axis=1)
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(subject_totals.index, subject_totals.values)
ax.set(xlabel="Subject", ylabel="Trials")
ax.tick_params(axis="x", rotation=75)
plt.show()
```

## Model-independent visualisations

```python
plot_class_waveforms(dataset)
plt.show()
plot_channel_effects(dataset)
plt.show()
```

## Reproducibility

The same task definition and cached preprocessing are used by every model. List the exact event mapping with `br-analysis list-tasks`; prepare caches with `br-analysis prepare-data`.

## Interpretation and limitations

Grand averages can hide subject heterogeneity and do not account for temporal dependence. These plots are descriptive only and must not be used to select subjects or tune a decoding model.
